```python
import h5py

file_path = r"8Q-t20-time-dephasing-target-heisenberg_20260628_115854\8Q-t20-time-dephasing-target-heisenberg_run000_8981107e-7a8f-413e-bcd5-9d2a5934349d_HISTORY.h5"

with h5py.File(file_path, "r") as f:
    print("Top-level objects:")
    for key in f.keys():
        print(key)
```

```python

with h5py.File(file_path, "r") as f:
    group = f["history_nit"]

    def print_name(name):
        print(name)

    group.visit(print_name)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

costs = []

with h5py.File(file_path, "r") as f:
    history = f["history_nfev"]

    # sort keys to ensure correct order
    keys = sorted(history.keys())

    for k in keys:
        cost = history[k]["cost"][()]   # scalar value
        costs.append(cost)

x = np.arange(len(costs))

plt.plot(x, costs)
plt.xlabel("Number function Evaluations")
plt.ylabel("Cost")
plt.title("Cost vs Number function Evaluations")
plt.grid(True)
plt.show()
```
