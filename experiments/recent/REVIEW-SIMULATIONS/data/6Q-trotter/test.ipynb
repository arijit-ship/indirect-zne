# 🔍 Experiment Validation

This notebook scans all experiment folders, checks each JSON file for `artifacts.opt_obj.success`, and reports:

- ✅ Successful optimizations
- ❌ Failed optimizations
- ⚠️ Missing keys
- 💥 Invalid or unreadable JSON files

A summary of the total, passed, and failed files is displayed at the end.

```python
from pathlib import Path
import json

# Root directory (change this to your target folder, or leave as cwd)
root = Path.cwd()
print(f"🚀 Recursively scanning '{root}' for *_VQE.json files...\n")

total = 0
passed = 0
failed = 0

# rglob searches through ALL subfolders, no matter how deep or what they're named
for json_file in sorted(root.rglob("*_VQE.json")):
    total += 1
    try:
        with open(json_file, "r") as f:
            data = json.load(f)
        success = data["artifacts"]["opt_obj"]["success"]
        if success:
            passed += 1
            print(f"✅ {json_file.relative_to(root)}  →  success: {success}")
        else:
            failed += 1
            print(f"❌ {json_file.relative_to(root)}  →  success: {success}")
    except KeyError as e:
        failed += 1
        print(f"⚠️  {json_file.relative_to(root)}  →  Missing key: {e}")
    except Exception as e:
        failed += 1
        print(f"💥 {json_file.relative_to(root)}  →  {type(e).__name__}: {e}")

print("\n" + "─" * 60)
print("📊 Scan Complete")
print(f"📄 Total *_VQE.json files : {total}")
print(f"✅ success = True         : {passed}")
print(f"❌ success = False/other  : {failed}")

if total == 0:
    print("\n⚠️ No *_VQE.json files found under this root.")
elif failed == 0:
    print("\n🎉 All optimization runs completed successfully!")
else:
    print(f"\n🚨 {failed} file(s) require attention.")
```
