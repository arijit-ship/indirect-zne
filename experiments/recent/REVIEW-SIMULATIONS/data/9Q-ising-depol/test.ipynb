# 🔍 Experiment Validation

This notebook scans all experiment folders, checks each JSON file for `artifacts.opt_obj.success`, and reports:

- ✅ Successful optimizations
- ❌ Failed optimizations
- ⚠️ Missing keys
- 💥 Invalid or unreadable JSON files

A summary of the total, passed, and failed files is displayed at the end.

```python
from pathlib import Path
import json

# Root directory (the folder containing this notebook)
root = Path.cwd()
print("🚀 Scanning experiment folders...\n")

total = 0
passed = 0
failed = 0

vqe_subfolders = ["VQE", "VQE_noiseoff"]

for vqe_name in vqe_subfolders:
    vqe_folder = root / vqe_name
    if not vqe_folder.is_dir():
        print(f"⚠️ no {vqe_name} subfolder found, skipping")
        continue

    print(f"📂 {vqe_name}")
    for json_file in sorted(vqe_folder.glob("*.json")):
        total += 1
        try:
            with open(json_file, "r") as f:
                data = json.load(f)
            success = data["artifacts"]["opt_obj"]["success"]
            if success:
                passed += 1
                print(f"   ✅ {json_file.name}")
            else:
                failed += 1
                print(f"   ❌ {json_file.name}  →  optimization failed")
        except KeyError as e:
            failed += 1
            print(f"   ⚠️  {json_file.name}  →  Missing key: {e}")
        except Exception as e:
            failed += 1
            print(f"   💥 {json_file.name}  →  {type(e).__name__}: {e}")
    print()  # blank line between folders

print("─" * 60)
print("📊 Scan Complete")
print(f"📄 Total JSONs : {total}")
print(f"✅ Passed      : {passed}")
print(f"❌ Failed      : {failed}")

if failed == 0:
    print("\n🎉 All optimization runs completed successfully!")
else:
    print(f"\n🚨 {failed} file(s) require attention.")
```
